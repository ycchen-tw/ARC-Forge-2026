```python
"""Reporting, staging bookkeeping, and the finders for every mounted input.

Forked on 2026-09-25 from work/arc_agi3/arc_code_qwen/notebook/01_prelude.py
(the old arm, which retires once this one has scored). Emitted as one cell by
work/arc_code/kaggle/build_notebook.py, which concatenates the files here in
name order.
"""


# Imported here for every cell: the notebook is one module split across cells,
# so a name this cell never uses may be the one the serve cell needs.
import base64
import datetime
import gzip
import hashlib
import io
import json
import os
import pathlib
import re
import shutil
import stat
import subprocess
import sys
import tarfile
import threading
import time
import urllib.request
import zipfile
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path

WORK = pathlib.Path("/kaggle/working")
# Everything the *agent* touches -- its venv, the coding CLI, arc-code's
# checkout and the run directory. None of it is evidence and none of it belongs
# in the commit: /kaggle/working is both the output and a 20 GB budget, and the
# venv, the CLI and the checkout used to be committed with it. The run directory
# is the one thing worth keeping and MAIN copies that back; MAIN's finally then
# removes this tree.
#
# It also has to be reachable when /kaggle is not -- the fence variant that hides
# the corpus does it by closing /kaggle to every unprivileged uid -- which is why
# this is not simply WORK.
#
# jail.sh gives the session a private tmpfs /tmp, so nothing here is visible
# inside the jail except what it binds through by absolute path: the venv,
# act.py, the codex binary and the game's own workspace. That was already true
# under /kaggle, which the jail does not bind at all, so the move does not widen
# or narrow what a game can see.
AGENT = pathlib.Path("/tmp/arcrun")
RUN = AGENT / "arm-run"
# The new arm is carried with its repo layout -- work/arc_code/ beside
# src/arcforge/ -- because run_arm.py finds the arcforge package at ../../src
# from itself. It lands under AGENT, not WORK: under the relocate fence /kaggle
# is closed to every game's uid, and codex, running as that uid, has to read the
# driver files (its base instructions and compaction prompt) out of this tree.
CODE = AGENT / "code"
UNIT = CODE / "work" / "arc_code"
T0 = time.monotonic()
# Background work that must not outlive a failure. ThreadPoolExecutor registers
# an atexit hook that joins its workers *before* daemon threads are killed, so a
# reader left running would hold the kernel open until it finished the file.
PENDING = []
state = {"stages": {}, "complete": False}


def save():
    (WORK / "kaggle-arm-state.json").write_text(json.dumps(state, indent=2))


def disk():
    usage = shutil.disk_usage(WORK)
    return f"{usage.free / 2**30:.1f}/{usage.total / 2**30:.1f} GiB free"


def say(*parts):
    """The only channel out of a running kernel is stdout.

    kaggle-arm-state.json records everything, but it cannot be read until the
    session ends, so a nine-hour run that printed only at the end would be
    unobservable for nine hours and would report a failure once, at the end.
    """
    print(f"[{time.strftime('%H:%M:%S')}]", *parts, flush=True)


def stage(name, fn):
    begin = time.monotonic()
    say("STAGE", name, "begin")
    try:
        value = fn()
    except BaseException as error:
        state["stages"][name] = {"ok": False, "error": f"{type(error).__name__}: {error}"}
        save()
        while PENDING:
            PENDING.pop()()
        say("STAGE", name, "FAILED", f"{type(error).__name__}: {error}")
        raise
    seconds = round(time.monotonic() - begin, 1)
    state["stages"][name] = {"ok": True, "seconds": seconds,
                             "value": value if isinstance(value, (str, int, float, list, dict)) else None}
    save()
    say("STAGE", name, f"ok in {seconds}s")
    return value


def run(argv, timeout, env=None, cwd=None, log=None):
    """Redirected output still has to reach the exception.

    With `log=` the child's stdout goes to a file and `done.stdout` is None, so
    the message used to read `... exited 1:` and stop. That file is only
    retrievable if the session ends cleanly -- which is not the case you need it
    in -- so the tail is read back and carried in the error.
    """
    path = WORK / (log or "stage")
    handle = path.open("a") if log else None
    try:
        done = subprocess.run(argv, timeout=timeout, env=env or os.environ.copy(), cwd=cwd,
                              stdout=handle or subprocess.PIPE,
                              stderr=subprocess.STDOUT, text=True)
    finally:
        if handle is not None:
            handle.close()
    output = done.stdout
    if output is None:
        output = path.read_text(errors="replace") if path.exists() else ""
    if done.returncode:
        raise RuntimeError(f"{argv[0]} exited {done.returncode}: {output[-4000:]}")
    return output[-2000:]


def mounted(name):
    """Find a mounted input by name rather than by an assumed path.

    A dataset can land under its own slug or under competitions/<slug>/, and
    Kaggle mounts some inputs as symlinks -- a glob that declines to follow one
    is indistinguishable from a dataset that never attached. Raises rather than
    returning None: every caller here needs what it asked for.
    """
    root = pathlib.Path("/kaggle/input")
    for depth in range(4):
        for entry in root.glob("/".join(["*"] * depth + [name])):
            return entry
    for parent, dirs, _ in os.walk(root, followlinks=True):
        if name in dirs:
            return pathlib.Path(parent) / name
        if parent.count("/") > 6:
            dirs[:] = []
    raise RuntimeError(f"{name} is not mounted under /kaggle/input; mounted: "
                       + ", ".join(sorted(p.name for p in root.iterdir())))


def mounted_glob(pattern):
    """The same lesson as mounted(), one level down: find a mounted *file*.

    A dataset's files sit at /kaggle/input/datasets/<owner>/<slug>/<file>, so the
    `*/<pattern>` this used to be matched nothing -- and paired with
    `next(..., None)` that turned a dataset which had attached perfectly into a
    silently different code path. Raises, and names what is actually mounted.
    Ambiguity is an error too: two matches mean the caller's pattern is wrong,
    and picking one of them would be a guess.
    """
    root = pathlib.Path("/kaggle/input")
    # Walked rather than globbed at a fixed depth. The official AOT dataset keeps
    # its library at datasets/<owner>/<slug>/official-sm120-aot/fused_moe_120/,
    # which is one level deeper than the five this used to search, and the glob
    # reported it as "not mounted" -- indistinguishable from a dataset that never
    # attached. os.walk also follows the symlinks Kaggle mounts some inputs as.
    hits = []
    for parent, dirs, files in os.walk(root, followlinks=True):
        if parent.count(os.sep) - str(root).count(os.sep) > 8:
            dirs[:] = []
            continue
        hits += [pathlib.Path(parent) / f for f in files
                 if pathlib.PurePath(f).match(pattern)]
    hits = sorted(hits)
    if len(hits) > 1:
        raise RuntimeError(f"{len(hits)} inputs match {pattern!r}: {hits}")
    if hits:
        return hits[0]
    raise RuntimeError(f"nothing matches {pattern!r} under /kaggle/input; mounted: "
                       + ", ".join(sorted(p.name for p in root.iterdir())))


def only(root, pattern):
    """One match or an error -- never a silently-chosen first of several."""
    hits = sorted(root.rglob(pattern))
    if len(hits) != 1:
        raise RuntimeError(f"expected exactly one {pattern!r} under {root}, found {hits}")
    return hits[0]


def unpack_arm():
    """Lay down the arm's own code, carried in the notebook rather than mounted.

    One deterministic gzipped tar; `filter="data"` refuses an absolute path or a
    `..` on the way out. Member names are repo-relative, so they -- not this
    function -- decide where things land, and that is checked rather than assumed.
    Everything is world-readable and nothing is world-writable: the games' uids
    read the driver files and fence scripts from here, and root executes them.
    """
    CODE.mkdir(parents=True, exist_ok=True)
    with tarfile.open(fileobj=io.BytesIO(gzip.decompress(base64.b64decode(PAYLOAD)))) as bundle:
        bundle.extractall(CODE, filter="data")
    for name in ("run_arm.py", "fence_posix.sh", "posix_fence.py"):
        if not (UNIT / name).is_file():
            raise RuntimeError(f"the carried bundle laid no {name} at {UNIT}")
    for path in [CODE, *CODE.rglob("*")]:
        path.chmod((0o755 if path.is_dir() or path.suffix == ".sh" else 0o644))
    return sorted(str(p.relative_to(CODE)) for p in CODE.rglob("*") if p.is_file())
```

```python
"""Lay down the agent's world: codex, the runtime, the two venvs.

Forked on 2026-09-25 from work/arc_agi3/arc_code_qwen/notebook/02_stage.py. What
changed is what the new arm needs: no arc-code checkout to patch (the new arm
ships its own act client), and **two** venvs instead of one -- the harness venv
holds arc_agi for the stand-in gateway, the agent venv holds only what an agent
may use. The old arm handed the agent the harness venv, and 19 of 25 games on
b200 read the actuator's source out of it.
"""

def shipped_pip():
    """pip has to be mounted, because this image can neither make one nor be given one.

    `python -m venv` runs ensurepip, and on this image ensurepip exits 1 --
    measured 2026-09-17. The competition's own arc_agi_3_wheels holds 31 wheels
    and pip is not among them. So one is shipped in its own dataset.
    """
    return mounted_glob("pip-*-py3-none-any.whl")


def sha256_of(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(16 << 20), b""):
            h.update(chunk)
    return h.hexdigest()


def stage_patched_codex(dataset, pinned, target):
    """The self-built codex (0.154.0 + patches/codex-0.154.0), copied off its mount and pinned.

    Checked three ways -- the config's pin, the dataset's own SHA256SUMS, and
    `--version` -- because a codex that is merely *a* 0.154.0 would run fine and
    silently lack view_video, the session release and the compaction prefix reuse.
    """
    root = mounted(dataset.split("/")[-1])
    native = only(root / "codex", "vendor/*/bin/codex")
    target = pathlib.Path(target)
    target.parent.mkdir(parents=True, exist_ok=True)
    shutil.copyfile(native, target)
    target.chmod(0o755)
    digest = sha256_of(target)
    if digest != pinned:
        raise RuntimeError(f"mounted codex is {digest}, the config pins {pinned}")
    recorded = (root / "codex/SHA256SUMS").read_text().split()[0]
    if recorded != digest:
        raise RuntimeError(f"the dataset's own SHA256SUMS says {recorded}, the file is {digest}")
    version = run([str(target), "--version"], 120).strip()
    if "0.154.0" not in version:
        raise RuntimeError(f"codex --version said {version!r}, not 0.154.0")
    return {"path": str(target), "sha256": digest, "version": version,
            "patches": (root / "codex/PATCHES").read_text().strip().splitlines()}


def stage_tools():
    """codex into AGENT/codexbin (the launcher's --driver-bin), and an arc-code prep.

    The prep is only *inspected* by the new arm (manifest.verify_prep records its
    commit and pins); it plays through its own act client. So it is copied as it
    comes, with no patches applied.
    """
    prefix = AGENT / "codexbin"
    codex = stage_patched_codex(CFG["codex_dataset"], CFG["codex_sha256"], prefix / "bin/codex")
    prep = AGENT / "arc-code-prep"
    if not prep.exists():
        shutil.copytree(mounted("arc-code-prep"), prep)
    for path in [AGENT, prefix, prefix / "bin"]:
        path.chmod(0o711 if path == AGENT else 0o755)
    return {**codex, "prefix": str(prefix), "prep": str(prep)}


def unpack_runtime():
    root = WORK / "runtime"
    if (root / "package").exists():
        return str(root / "package")
    archive = mounted_glob("*sm120-cp312.tar.zst")
    zstd = mounted_glob("zstd")
    root.mkdir(parents=True, exist_ok=True)
    # That dataset ships liblz4/liblzma/libz beside the binary because the
    # image does not have all three; copying only `zstd` gives a loader error.
    for lib in zstd.parent.iterdir():
        shutil.copy(lib, WORK / lib.name)
    (WORK / "zstd").chmod(0o755)
    run(["bash", "-c", f"{WORK}/zstd -d -c {archive} | tar -x -C {root}"], 1800,
        env=dict(os.environ, LD_LIBRARY_PATH=f"{WORK}:" + os.environ.get("LD_LIBRARY_PATH", "")))
    return str(root / "package")


def find_model():
    """Locate the weights by what they contain, not by a guessed mount path."""
    hint = CFG.get("model_hint", "")
    found = []
    for parent, dirs, files in os.walk("/kaggle/input", followlinks=True):
        if "config.json" in files and any(f.endswith(".safetensors") for f in files):
            found.append(parent)
            dirs[:] = []
    if not found:
        raise RuntimeError("no checkpoint under /kaggle/input: "
                           + ", ".join(sorted(p.name for p in pathlib.Path("/kaggle/input").iterdir())))
    matching = [p for p in found if hint in p] or found
    if len(matching) > 1:
        raise RuntimeError(f"{len(matching)} checkpoints match {hint!r}: {matching}")
    # Model versions that share config.json (QAD v2 of the gEAN320 INT3 package) differ in deploy_manifest.json,
    # which lists every shard's sha256: pinning it pins the weights without hashing 89 GB here.
    pinned = CFG.get("model_manifest_sha256")
    if pinned is not None:
        got = sha256_of(pathlib.Path(matching[0]) / "deploy_manifest.json")
        if got != pinned:
            raise RuntimeError(f"{matching[0]}/deploy_manifest.json is {got}, not the pinned {pinned}")
    if CFG.get("model_overlay"):
        return overlay_model(pathlib.Path(matching[0]), CFG["model_overlay"], OVERLAY_MODEL_DIR)
    return matching[0]


#: config ``model_overlay``: the served checkpoint, a directory of symlinks (outside /kaggle/working, which is output)
OVERLAY_MODEL_DIR = pathlib.Path("/tmp/arc-model-overlay")


def overlay_model(base, spec, dest):
    """The mounted package ``base`` (pinned by model_manifest_sha256) with some shards replaced by an overlay's
    (work/qwen38fn_rl/rl_overlay_package.py: the shards an RL version rewrote, its deploy_manifest.json and
    overlay.json), from a dataset or a model (overlay_root). Kaggle stores ~20 GB instead of a second 89 GB package; the
    served directory is symlinks, so no large write on a machine that vanishes past 80-90 GiB of writes. Raises unless
    the overlay's manifest is the pinned one and was made against this base, it differs from the base manifest in
    exactly the overlay's shards, and each of those has its manifest's size. Sizes, not sha256: v30 hashed them here
    and the cold read (~31 MB/s) took 649 s of the rerun's play time; the bytes are pinned by the overlay manifest's
    sha256 (checked here) and the immutable dataset / model version the kernel mounts."""
    root = overlay_root(spec)
    overlay = json.loads((root / "overlay.json").read_text())
    if overlay["base_manifest_sha256"] != CFG["model_manifest_sha256"]:
        raise RuntimeError(f"the overlay was made against {overlay['base_manifest_sha256']}, "
                           f"the mounted package is {CFG['model_manifest_sha256']}")
    got = sha256_of(root / "deploy_manifest.json")
    if not got == spec["manifest_sha256"] == overlay["deploy_manifest_sha256"]:
        raise RuntimeError(f"overlay deploy_manifest.json is {got}, pinned {spec['manifest_sha256']}, "
                           f"overlay.json says {overlay['deploy_manifest_sha256']}")
    base_files = json.loads((base / "deploy_manifest.json").read_text())["files"]
    new_files = json.loads((root / "deploy_manifest.json").read_text())["files"]
    replaced = {name for name, entry in new_files.items() if base_files.get(name) != entry}
    if set(new_files) != set(base_files) or replaced != set(overlay["files"]):
        raise RuntimeError(f"overlay manifest: files {sorted(set(new_files) ^ set(base_files))} differ in name, "
                           f"replaced {sorted(replaced)} vs overlay.json {sorted(overlay['files'])}")
    for name in sorted(replaced):
        path = root / name
        if path.stat().st_size != new_files[name]["bytes"]:
            raise RuntimeError(f"{path} is {path.stat().st_size} bytes, the overlay manifest says "
                               f"{new_files[name]['bytes']}")
    links = {entry.name: entry for entry in base.iterdir()}
    links.update({name: root / name for name in [*sorted(replaced), "deploy_manifest.json"]})
    dest.mkdir(parents=True)
    for name, target in sorted(links.items()):
        (dest / name).symlink_to(target)
    say("MODEL OVERLAY", f"{dest}: {len(links)} links, {len(replaced)} shards from {root}")
    return str(dest)


def overlay_root(spec):
    """Where overlay.json is: a dataset (``dataset``) mounts its files at its root, a model (``model``, the handle
    owner/slug/framework/variation/version) a framework/variation/version deeper."""
    if "dataset" in spec:
        return mounted(spec["dataset"].split("/")[-1])
    return only(mounted(spec["model"].split("/")[1]), "overlay.json").parent


def make_venv(name, wheels, packages, check, extra_links=()):
    """A venv without ensurepip, filled from the competition wheels with the pip we mounted.

    --without-pip is the whole point rather than a fallback: asking for a pip is
    what killed the old arm's v4. A wheel is a zip with pip/__main__.py inside it,
    so `python <wheel>/pip` runs pip without installing it; running it with the
    *venv's* interpreter is what makes it install into the venv.
    """
    venv = AGENT / name
    base = "/usr/bin/python3.12" if pathlib.Path("/usr/bin/python3.12").exists() else sys.executable
    state.setdefault("venv_base", base)
    if not (venv / "bin/python").exists():
        run([base, "-m", "venv", "--without-pip", str(venv)], 300)
    pip = shipped_pip()
    state["pip_wheel"] = pip.name
    save()
    links = [a for d in (wheels, *extra_links) for a in ("--find-links", str(d))]
    run([str(venv / "bin/python"), f"{pip}/pip", "install", "--no-index", *links,
         "--disable-pip-version-check", "-q", *packages], 900)
    run([str(venv / "bin/python"), "-c", check], 120, env=dict(os.environ, MPLBACKEND="Agg"))
    return str(venv / "bin/python")


def stage_agent_wheels(spec):
    """Config ``agent_wheels``: wheels the competition's arc_agi_3_wheels lacks (scipy), from our own dataset.

    Each file is found by name, checked against the config's sha256 and copied to one
    directory that make_venv adds as a second --find-links; the agent venv's other
    packages still come from the competition's wheels. RL's jail runtime
    ($ARC_RUNS/arc_code/agent-runtime-scipy) is the same wheel files built the same way.
    """
    dest = AGENT / "agent-wheels"
    dest.mkdir(parents=True, exist_ok=True)
    for name, want in spec["files"].items():
        source = mounted_glob(name)
        got = sha256_of(source)
        if got != want:
            raise RuntimeError(f"{source} is {got}, the config pins {want}")
        shutil.copyfile(source, dest / name)
    state["agent_wheels"] = dict(spec["files"])
    save()
    return dest


def make_venvs(wheels):
    """harness: arc_agi + flask, for the stand-in gateway (root only).
    agent: numpy, requests, pillow, matplotlib (+ config ``agent_wheels``, e.g. scipy) -- and provably
    no arcengine, because fence_posix.sh refuses a session whose runtime can import it."""
    harness = make_venv("harness-venv", wheels, ["arc-agi==0.9.8", "flask"],
                        "from arc_agi import Arcade; from arc_agi.server import create_app")
    packages, links, check = ["numpy", "requests", "pillow", "matplotlib"], [], "import numpy, requests, PIL, matplotlib; "
    if CFG.get("agent_wheels"):
        spec = CFG["agent_wheels"]
        links = [stage_agent_wheels(spec)]
        packages += spec["packages"]
        check += "".join(f"import {p.split('==')[0]}; assert {p.split('==')[0]}.__version__ == {p.split('==')[1]!r}; "
                         for p in spec["packages"]) + spec["check"] + "; "
    agent = make_venv("agent-venv", wheels, packages,
                      check + "import importlib.util as u; raise SystemExit(u.find_spec('arcengine') is not None)", links)
    # The agent venv is read by every game's uid; the harness venv by root only.
    run(["chmod", "-R", "a+rX,go-w", str(AGENT / "agent-venv")], 300)
    (AGENT / "harness-venv").chmod(0o700)
    return {"harness": harness, "agent": agent}
```

```python
"""The page-cache warm, lifted verbatim from work/qwen38fn_eval/sm120_runtime.py."""


def prefetch_checkpoint(source: Path, workers: int = 16,
                        chunk_bytes: int = 256 << 20) -> dict:
    """Warm input-mount page cache without duplicating 95 GB on the root disk.

    Explicit reads finish before the loader starts. Per-range SHA256 records
    exactly what was prefetched; residency is best-effort under kernel eviction.
    """
    started = time.monotonic()
    shards = sorted(source.glob("*.safetensors"))
    if not shards or workers < 1:
        raise ValueError("Expected checkpoint shards and a positive worker count")
    total = sum(p.stat().st_size for p in shards)
    jobs = [(p, offset, min(chunk_bytes, p.stat().st_size - offset))
            for p in shards for offset in range(0, p.stat().st_size, chunk_bytes)]

    def read_range(job):
        path, offset, length = job
        digest = hashlib.sha256()
        with path.open("rb") as stream:
            stream.seek(offset)
            remaining = length
            while remaining:
                block = stream.read(min(remaining, 8 << 20))
                if not block:
                    raise EOFError(f"Short prefetch read: {path}:{offset}")
                digest.update(block)
                remaining -= len(block)
        return {"file": path.name, "offset": offset, "bytes": length,
                "sha256": digest.hexdigest()}

    ranges, loaded, reported = [], 0, started
    with ThreadPoolExecutor(max_workers=workers) as pool:
        for future in as_completed([pool.submit(read_range, job) for job in jobs]):
            row = future.result()
            ranges.append(row)
            loaded += row["bytes"]
            if time.monotonic() - reported >= 15:
                reported = time.monotonic()
                print(f"CHECKPOINT_PREFETCH {loaded / 1e9:.1f}/{total / 1e9:.1f} GB "
                      f"elapsed={reported - started:.1f}s threads={workers}", flush=True)
    if loaded != total:
        raise RuntimeError("Prefetch coverage is incomplete")
    seconds = time.monotonic() - started
    return {"mode": "pagecache", "source": str(source), "workers": workers,
            "bytes": total, "seconds": seconds, "GB_per_second": total / seconds / 1e9,
            "ranges": sorted(ranges, key=lambda r: (r["file"], r["offset"]))}
```

```python
"""Bring up sglang on the pinned runtime, and prove it is serving what was asked for.

Forked verbatim on 2026-09-25 from work/arc_agi3/arc_code_qwen/notebook/03_serve.py:
the runtime, the rewarm and the startup checks are the parts of the old arm that
Kaggle has already proven, and nothing about the new arm changes how sglang starts.
"""


def runtime_interpreter():
    """The pinned runtime is a cp312 site-packages tree, not a venv, so it needs a
    system interpreter of exactly that version. Kaggle's image python is 3.12,
    but it is not necessarily at /usr/bin/python3.12.
    """
    interpreter = "/usr/bin/python3.12" if pathlib.Path("/usr/bin/python3.12").exists() else sys.executable
    if subprocess.run([interpreter, "-c", "import sys; raise SystemExit(sys.version_info[:2] != (3, 12))"]).returncode:
        raise RuntimeError(f"no python3.12 for the cp312 runtime; tried {interpreter}")
    return interpreter


def runtime_env(runtime, env, arch_list=None):
    """The environment this runtime is served with everywhere else, not a new guess.

    v5 built its own env here holding nothing but a PYTHONPATH, and sglang died
    inside flashinfer's JIT with "FlashInfer requires GPUs with sm75 or higher"
    on an sm120 card. That message is what an *empty* TARGET_CUDA_ARCHS prints:
    the set is filled from the CUDA toolchain, and nothing here was pointing at
    one. The archive ships its own (manifest `layout.cuda` = package/cuda-13.1),
    which is exactly what work/qwen38fn_eval/qsa_sm120_pipeline.py's
    runtime_environment() -- the env that has actually served this tree on this
    image -- puts on PATH. Mirroring it is the point; the tools it asserts on are
    the ones flashinfer shells out to.
    """
    root = pathlib.Path(runtime)
    site = root / "lib/python3.12/site-packages"
    env = dict(env,
               CUDA_HOME=str(root / "cuda-13.1"),
               CUDA_VISIBLE_DEVICES="0",
               PATH=os.pathsep.join([str(root / "cuda-13.1/bin"), str(root / "bin"), env["PATH"]]),
               PYTHONPATH=os.pathsep.join([str(root / "python"), str(site),
                                           str(site / "nvidia_cutlass_dsl/dsl_packages")]),
               MAX_JOBS="2", FLASHINFER_NVCC_THREADS="1", MPLBACKEND="Agg")
    # flashinfer decides what to compile for from TARGET_CUDA_ARCHS, and left to
    # itself it derives that through a CUDA-version probe that can fail quietly:
    # every failure inside CompilationContext.__init__ is swallowed into a
    # warning and leaves the set *empty*, which check_cuda_arch then reports as
    # "requires GPUs with sm75 or higher" -- the v5 death, on an sm120 card.
    # A value carrying an arch suffix is taken as-is, so passing the one measured
    # off this GPU skips that path entirely instead of hoping it works.
    if arch_list:
        env["FLASHINFER_CUDA_ARCH_LIST"] = arch_list
    missing = [tool for tool in ("nvcc", "ninja", "g++")
               if shutil.which(tool, path=env["PATH"]) is None]
    if missing:
        raise RuntimeError(f"the pinned runtime cannot compile: {missing} not on {env['PATH']}")
    return env


CUDA_REPORT = r"""
import json, os
report = {}
def note(name, fn):
    try:
        report[name] = fn()
    except BaseException as error:
        report[name] = f"{type(error).__name__}: {error}"
import torch
note("torch", lambda: torch.__version__)
note("torch_cuda", lambda: torch.version.cuda)
note("device_count", lambda: torch.cuda.device_count())
note("capability", lambda: list(torch.cuda.get_device_capability(0)))
report["env_arch_list"] = {name: os.environ.get(name) for name in
                           ("FLASHINFER_CUDA_ARCH_LIST", "TORCH_CUDA_ARCH_LIST")}
from flashinfer.jit.cpp_ext import get_cuda_path, get_cuda_version
note("cuda_path", get_cuda_path)
note("cuda_version", lambda: str(get_cuda_version()))
from flashinfer.compilation_context import CompilationContext
# Its own normalisation, not a copy of it: the sm12.x suffix rule lives there and
# changes with the CUDA version it detects.
note("arch_list", lambda: "%d.%s" % CompilationContext._normalize_cuda_arch(
    *torch.cuda.get_device_capability(0)))
note("target_archs", lambda: sorted(f"{a}.{b}" for a, b in CompilationContext().TARGET_CUDA_ARCHS))
print(json.dumps(report))
"""


def eligible(target_archs):
    """flashinfer's own rule, applied where the numbers are still readable.

    check_cuda_arch accepts sm75 and up, so an empty list is not the only way to
    fail it: a list holding nothing but older architectures -- which is what an
    inherited FLASHINFER_CUDA_ARCH_LIST would produce -- fails it too, and prints
    the same sentence about sm75.
    """
    if not isinstance(target_archs, list):
        return False
    for arch in target_archs:
        major, _, minor = str(arch).partition(".")
        if major.isdigit() and (int(major) >= 8 or (major == "7" and minor[:1].isdigit()
                                                    and int(minor[:1]) >= 5)):
            return True
    return False


def cuda_report(interpreter, env):
    """Measure what flashinfer will decide, before sglang spends ten minutes on it.

    Every field is recorded rather than asserted, except the one that decides:
    an empty arch list is the v5 failure, and it is worth naming here -- where
    the numbers that produced it are still in hand -- instead of one stage later
    as a sentence about sm75.
    """
    done = subprocess.run([interpreter, "-c", CUDA_REPORT], capture_output=True, text=True,
                          timeout=600, env=env)
    if done.returncode:
        raise RuntimeError(f"cuda probe exited {done.returncode}: {(done.stdout + done.stderr)[-1200:]}")
    report = json.loads(done.stdout.strip().splitlines()[-1])
    # One rule, flashinfer's own, applied to the value actually being pinned.
    # Every field here is produced by a helper that turns failures into strings,
    # so "is a string" proves nothing -- "would satisfy check_cuda_arch" does.
    if not eligible([report.get("arch_list")]):
        raise RuntimeError("no usable architecture to pin for flashinfer, which is the v5 death "
                           "dressed as a message about sm75: " + json.dumps(report))
    return report


def assert_runtime_provenance(runtime):
    """The patch stack is already in the archive; this is where that is checked.

    The old line applied patches here, on Kaggle, which meant "one patch did not
    apply" was discovered in the one place nobody can look -- and it happened:
    an installer's output piped into `head` gave `patch` a SIGPIPE and left the
    stack half-applied. package_runtime_sm120_0520.sh now runs the marker gate
    on the b200 and writes runtime-provenance.json beside the tree, so all that
    is left here is to refuse an archive that is not the one that was gated.

    Hashes, not markers: a marker says a patch was applied, a hash says the file
    is the file. The archive could still be an older build of the same stack.
    """
    root = pathlib.Path(runtime)
    record_path = root / "runtime-provenance.json"
    if not record_path.exists():
        raise RuntimeError(f"{record_path} is missing: this archive was not built by "
                           "package_runtime_sm120_0520.sh, so nothing knows what is in it")
    record = json.loads(record_path.read_text())
    expected = CFG.get("runtime_source_sha256") or {}
    if not expected:
        raise RuntimeError("the config carries no runtime_source_sha256, so the archive "
                           "cannot be pinned; add it from the packager's provenance")
    actual = record.get("source_sha256") or {}
    drifted = sorted(name for name, digest in expected.items() if actual.get(name) != digest)
    if drifted:
        raise RuntimeError(f"the mounted runtime is not the one this notebook was built "
                           f"against; {len(drifted)} files differ: {drifted[:6]}")
    missing = sorted(set(CFG.get("runtime_patches") or []) - set(record.get("patches") or {}))
    if missing:
        raise RuntimeError(f"the archive was built without {missing}")
    say("RUNTIME", json.dumps({"base_ref": record.get("base_ref"),
                               "arc_forge_commit": record.get("arc_forge_commit"),
                               "patches": len(record.get("patches") or {}),
                               "pinned_files": len(expected),
                               "aot": (record.get("aot") or {}).get("modules")}))
    return {"base_ref": record.get("base_ref"),
            "arc_forge_commit": record.get("arc_forge_commit"),
            "patches": sorted(record.get("patches") or {}),
            "aot_bytes": (record.get("aot") or {}).get("bytes")}


def apply_runtime_overlay(runtime, entries, dataset):
    """Replace pinned runtime files with pinned files shipped in `dataset`.

    How one sglang file changes without repackaging the multi-GB runtime: the
    arm uses it for the rewritten 0018 (serving_responses.py), which the runtime
    dataset does not have yet. Runs after assert_runtime_provenance(), so the
    archive underneath is still the one runtime_source_sha256 pins -- that check
    is about the archive and is left exactly as it was. Every entry is pinned
    three ways (the file it replaces, the shipped file, the file afterwards),
    `dest` must be the copy Python imports -- the first match along the
    runtime's PYTHONPATH -- and every mismatch raises.
    """
    if not entries:
        return "none"
    root = pathlib.Path(runtime)
    shipped_root = mounted(dataset.split("/")[-1])
    search = [pathlib.Path(p) for p in runtime_env(runtime, os.environ.copy())["PYTHONPATH"].split(os.pathsep)]
    applied = []
    for entry in entries:
        dest, source = root / entry["dest"], shipped_root / entry["source"]
        pinned = CFG["runtime_source_sha256"].get(entry["dest"])
        if pinned != entry["replaces_sha256"]:
            raise RuntimeError(f"overlay {entry['dest']} replaces {entry['replaces_sha256']}, but "
                               f"runtime_source_sha256 pins {pinned}")
        module = pathlib.Path(entry["dest"]).relative_to("python")
        imported = next((d / module for d in search if (d / module).is_file()), None)
        if imported != dest:
            raise RuntimeError(f"Python would import {module} from {imported}, not the overlay target {dest}")
        before = sha256_of(dest)
        if before != entry["replaces_sha256"]:
            raise RuntimeError(f"{dest} is {before}, the overlay expects to replace {entry['replaces_sha256']}")
        shipped = sha256_of(source)
        if shipped != entry["source_sha256"]:
            raise RuntimeError(f"{source} is {shipped}, the config pins {entry['source_sha256']}")
        shutil.copyfile(source, dest)
        after = sha256_of(dest)
        if after != entry["result_sha256"]:
            raise RuntimeError(f"{dest} is {after} after the copy, the config pins {entry['result_sha256']}")
        stale = sorted((dest.parent / "__pycache__").glob(dest.stem + ".*.pyc"))
        for pyc in stale:
            pyc.unlink()
        applied.append({"dest": entry["dest"], "source": entry["source"], "before": before, "after": after,
                        "stale_pyc_removed": [p.name for p in stale]})
        say("OVERLAY", entry["dest"], before[:12], "->", after[:12], f"{len(stale)} stale pyc removed")
    return applied


#: serve_e2e.sh's FORBIDDEN list: test-only switches (determinism, debug checks, skipped I/O). Any of them in a
#: served environment changes what is measured without saying so.
TEST_ONLY_ENV = ("ARC_HISPARSE_QSA_SYNC_STAGING", "ARC_QSA_TOPK_EXACT", "ARC_HC_MIX_DETERMINISTIC",
                 "SGLANG_FLASHINFER_MOE_FUSED_FINALIZE", "ARC_QSA_TOPK_SORT", "ARC_HISPARSE_QSA_DEBUG",
                 "SGLANG_DEBUG_HISPARSE_SKIP_IO")

#: distributions() walks sys.path in order and also yields the image's own copies further down (Kaggle's torch
#: 2.10.0 behind the runtime's 2.13.0), so keep the first of each name: that is the one import resolves.
TREE_PROBE = r"""
import importlib.metadata as m, importlib.util as u, json
want = {"torch", "triton", "flashinfer-python", "flashinfer-jit-cache", "sglang-kernel", "sgl-kernel", "tilelang"}
versions = {}
for d in m.distributions():
    if d.metadata["Name"] in want:
        versions.setdefault(d.metadata["Name"], d.version)
print(json.dumps({"origin": u.find_spec("sglang").origin, "versions": versions}))
"""


def install_sglang_tree(runtime):
    """Replace the archive's sglang source with the HiSparse tree, and prove the server will import that tree.

    The runtime archive stays the one v12 serves (assert_runtime_provenance has just checked it): its
    site-packages, the flashinfer AOT set and the CUDA toolchain are what the G4 runs used too. Only its
    python/sglang (0001-0028 and the old 0018) goes, replaced by v0.5.20 + 0001-0029 + experimental-hisparse
    h*, built and asserted on the PC (work/qsa_hisparse/probes/pack_kaggle_tree.py) -- the G4 fast path does
    the same (g4_runtime_install.sh). Pinned three ways: the archive's sha256, the whole-package digest after
    extraction (patches/sglang-0.5.20/TREE_DIGEST, arcforge.sglang_tree), and the import origin as the runtime
    interpreter sees it. The manifest h05 checks at startup comes from the same dataset, pinned too.
    """
    spec = CFG["sglang_tree"]
    root = pathlib.Path(runtime)
    archive = mounted_glob(spec["archive"])
    manifest = archive.parent / spec["manifest"]
    for path, pinned in ((archive, spec["archive_sha256"]), (manifest, spec["manifest_sha256"])):
        got = sha256_of(path)
        if got != pinned:
            raise RuntimeError(f"{path} is {got}, the config pins {pinned}")
    zstd = WORK / "zstd"
    if not zstd.is_file():
        raise RuntimeError(f"no {zstd}: unpack_runtime lays it down, and it has to run first")
    plain = WORK / "sglang-tree.tar"
    run(["bash", "-c", f"{zstd} -q -d -c {archive} > {plain}"], 600,
        env=dict(os.environ, LD_LIBRARY_PATH=f"{WORK}:" + os.environ.get("LD_LIBRARY_PATH", "")))
    with tarfile.open(plain) as tar:
        members = tar.getmembers()
        outside = [m.name for m in members if m.name != "python/sglang" and not m.name.startswith("python/sglang/")]
        if outside:
            raise RuntimeError(f"the tree archive has members outside python/sglang/: {outside[:5]}")
        shutil.rmtree(root / "python")
        tar.extractall(root, filter="data")
    plain.unlink()
    if str(CODE / "src") not in sys.path:
        sys.path.insert(0, str(CODE / "src"))
    from arcforge.sglang_tree import manifest_digest, package_manifest
    entries = package_manifest(root / "python" / "sglang")
    digest = manifest_digest(entries)
    if (digest, len(entries)) != (spec["tree_digest"], spec["tree_files"]):
        raise RuntimeError(f"the extracted tree is {digest} ({len(entries)} files), the config pins "
                           f"{spec['tree_digest']} ({spec['tree_files']} files)")
    done = subprocess.run([runtime_interpreter(), "-c", TREE_PROBE], capture_output=True, text=True, timeout=300,
                          env=runtime_env(runtime, os.environ.copy()))
    if done.returncode:
        raise RuntimeError(f"tree probe exited {done.returncode}: {(done.stdout + done.stderr)[-1200:]}")
    probe = json.loads(done.stdout.strip().splitlines()[-1])
    expected = root / "python" / "sglang" / "__init__.py"
    if pathlib.Path(probe["origin"]) != expected:
        raise RuntimeError(f"the runtime interpreter imports sglang from {probe['origin']}, not {expected}")
    report = {"digest": digest, "files": len(entries), "archive": str(archive), "manifest": str(manifest),
              "members": len(members), "versions": probe["versions"]}
    say("SGLANG TREE", json.dumps(report))
    return report


def cgroup_now():
    """The container's memory as the kernel accounts it (GiB), with the headroom posix_fence.py watches:
    memory.max - anon - shmem. Page cache is left out on purpose -- the kernel reclaims it before it kills
    anything, and the notebook keeps the checkpoint in it deliberately (rewarm). Pinned host memory (the PLE
    table, the HiSparse host pool) is shmem. Without a ceiling, MemAvailable is the headroom."""
    rel = Path("/proc/self/cgroup").read_text().strip().splitlines()[-1].split(":", 2)[2]
    base = Path("/sys/fs/cgroup") / rel.lstrip("/")
    stat = dict(line.split() for line in (base / "memory.stat").read_text().splitlines())
    gib = {k: int(stat[k]) / 2**30 for k in ("anon", "shmem", "file")}
    ceiling = (base / "memory.max").read_text().strip()
    if ceiling == "max":
        headroom = next(int(line.split()[1]) for line in Path("/proc/meminfo").read_text().splitlines()
                        if line.startswith("MemAvailable:")) / 2**20
    else:
        headroom = int(ceiling) / 2**30 - gib["anon"] - gib["shmem"]
    return {**{k: round(v, 1) for k, v in gib.items()}, "headroom": round(headroom, 1), "max": ceiling}


class MemorySampler(threading.Thread):
    """cgroup memory every `every` s until stopped: the low point of the headroom and the peaks, e.g. while
    the server starts (a cold JIT cache compiles then, next to the checkpoint in the page cache)."""

    def __init__(self, every=2.0):
        super().__init__(daemon=True, name="memory-sampler")
        self.every, self.halt, self.samples = every, threading.Event(), []

    def run(self):
        while True:
            self.samples.append(cgroup_now())
            if self.halt.wait(self.every):
                return

    def stop(self):
        self.halt.set()
        self.join(timeout=30)
        if not self.samples:
            raise RuntimeError("the memory sampler took no sample")
        low = min(self.samples, key=lambda s: s["headroom"])
        return {"samples": len(self.samples), "headroom_min": low["headroom"], "at_headroom_min": low,
                **{f"{k}_max": max(s[k] for s in self.samples) for k in ("anon", "shmem", "file")}}


def jit_cache_dirs(env):
    """Where anything compiled at startup lands: the cache directories the env names, and the defaults under
    $HOME that the runtime also writes (tilelang, triton, the CUDA cache)."""
    home = pathlib.Path(env.get("HOME") or os.path.expanduser("~"))
    named = [env[k] for k in ("SGLANG_CACHE_DIR", "SGLANG_JIT_CACHE_DIR", "TRITON_CACHE_DIR", "CUDA_CACHE_PATH",
                              "FLASHINFER_WORKSPACE_BASE") if env.get(k)]
    return sorted({pathlib.Path(p) for p in named}
                  | {home / ".cache", home / ".triton", home / ".nv", home / ".tilelang"})


def cache_files(dirs):
    return {str(p) for d in dirs if d.is_dir() for p in d.rglob("*") if p.is_file()}


def hisparse_startup_checks(text):
    """What the HiSparse server log must say beyond the static require_log markers: the tree h05 verified
    (the manifest's sha256), and the sizing R, C, L this config asks for. B (hot blocks per request) is read,
    reported against the G4 run, and not enforced: it follows from the VRAM left after the weights, which is
    a measurement, not a contract; h17 refuses B < K x top-k by itself."""
    spec = CFG["sglang_tree"]
    args = CFG["server_args"]
    logical = json.loads(args[args.index("--hisparse-config") + 1])["qsa_logical_tokens"]
    for marker in (f"ARC_HISPARSE_QSA_TREE={spec['manifest_sha256']}",
                   f"R={CFG['max_requests']}, C={CFG['context_length']} -> logical {logical} tokens"):
        if marker not in text:
            near = [line[-300:] for line in text.splitlines() if "QSA HiSparse sizing" in line or
                    "ARC_HISPARSE_QSA_TREE" in line][:4]
            raise RuntimeError(f"server log never said {marker!r}; nearby: {near or 'none'}")
    sizing = re.search(r"QSA HiSparse sizing: budget ([\d.]+) GiB, R=(\d+), C=(\d+) -> logical (\d+) tokens, "
                       r"prefill area (\d+) tokens, B=(\d+) blocks", text)
    if sizing is None:
        raise RuntimeError("the server log has the sizing marker but no parsable sizing line")
    keep = ("QSA HiSparse host pool:", "QSA HiSparse GDN snapshots:", "QSA HiSparse session host budget:",
            "QSA HiSparse draft pool:", "available_gpu_mem=")
    report = {"budget_gib": float(sizing[1]), "R": int(sizing[2]), "C": int(sizing[3]), "L": int(sizing[4]),
              "prefill_area": int(sizing[5]), "B": int(sizing[6]), "B_g4_c55": CFG.get("expect_hot_blocks"),
              "lines": [line[line.index(k):][:240] for line in text.splitlines() for k in keep if k in line][:8]}
    say("HISPARSE", json.dumps(report))
    return report


def start_prefetch(model):
    """Warm the checkpoint's page cache while the runtime is still unpacking.

    sglang's loader mmaps the shards, and mmap is only fast on a warm cache --
    that is the pairing `vista-sequential-nvfp4/config.json` records as "mmap
    loader follows verified page-cache warming". Reading them here, on 16
    threads, overlaps the cold NFS read with stages that are waiting on CPU and
    disk anyway. It is not sufficient on its own: see rewarm().
    """
    outcome = {}

    def work():
        begin = time.monotonic()
        try:
            outcome["manifest"] = prefetch_checkpoint(Path(model), CFG.get("prefetch_workers", 16))
        except BaseException as error:  # a cold cache is slower, not fatal
            outcome["error"] = f"{type(error).__name__}: {error}"
        outcome["seconds"] = round(time.monotonic() - begin, 1)

    thread = threading.Thread(target=work, daemon=True)
    thread.start()
    say("PREFETCH", "started on", model, disk())
    handle = (thread, outcome)
    PENDING.append(lambda: join_prefetch(handle))
    return handle


def join_prefetch(handle):
    thread, outcome = handle
    thread.join()
    manifest = outcome.get("manifest") or {}
    report = {"seconds": outcome.get("seconds"), "error": outcome.get("error"),
              "bytes": manifest.get("bytes"), "GB_per_second": manifest.get("GB_per_second")}
    say("PREFETCH", json.dumps(report))
    if outcome.get("error"):
        # Named, not raised: the loader will simply read it cold instead.
        state["prefetch_failed"] = outcome["error"]
    return report


def rewarm(model):
    """Read the checkpoint a second time, right before the server, so it survives the load.

    One read leaves every page on the inactive LRU list, and the server's own
    ~75 GiB of host memory pushes the 175 GiB cgroup to its limit while the
    loader runs -- so the kernel reclaims exactly those pages and the mmap
    loader refaults them off NFS. Measured 2026-09-24 (startup probe v1): the
    arm's single prefetch, 100% resident at launch, still refaulted 106.9 GB
    and took 1,174 s in Load weight; a second read, which promotes the pages
    to the active list, brought it to 31 s with 1.1 GB refaulted. The second
    read itself costs ~4 s because it is served from the cache it protects.

    It is a warm read only if it is fast. A slow one means the first pass was
    already evicted; that is recorded and printed, not raised -- the server
    still starts, just at the old speed.
    """
    begin = time.monotonic()
    manifest = prefetch_checkpoint(Path(model), CFG.get("prefetch_workers", 16))
    report = {"seconds": round(time.monotonic() - begin, 1),
              "GB_per_second": round(manifest["GB_per_second"], 3)}
    if report["GB_per_second"] < 5:
        state["rewarm_cold"] = report
        say("REWARM", "COLD: the second read went to NFS, so the first was already evicted "
            "and the load will be slow:", json.dumps(report))
    else:
        say("REWARM", json.dumps(report))
    return report


def dflash_env():
    """ARC_DFLASH_* for the MTP5-DFlash drafter (sglang-arc-2 speculative/arc_dflash), resolved on the mounted
    dataset: the export directory, its keep set and GEMM2 calibration, each pinned by sha256 (the export by its
    config.json; the worker itself refuses an incomplete export). A dense export (`ffn_type: dense_shared`,
    arc-forge convert_dense.py) has no experts to keep or calibrate: only its config is pinned, and its
    config.json must say dense_shared; plan B's plain / injected-context export (`ffn_type: dense`) likewise.
    An optional draft vocabulary (draft_vocab*) restricts the drafter's head to
    a pinned id list. The worker replaces the native MTP draft, so the checkpoint's MTP weights are not loaded and
    there is no draft KV pool."""
    spec = CFG["dflash"]
    dense = spec.get("ffn_type") in ("dense_shared", "dense")
    root = mounted(spec["dataset"].split("/")[-1])
    export = root / spec["export_dir"] if spec["export_dir"] else root
    files = {"config": (export / "config.json", spec["config_sha256"])}
    if not dense:
        files |= {"keep": (root / spec["keep"], spec["keep_sha256"]),
                  "calib": (root / spec["calib"], spec["calib_sha256"])}
    for name, (path, pinned) in files.items():
        got = sha256_of(path)
        if got != pinned:
            raise RuntimeError(f"dflash {name} {path} is {got}, the config pins {pinned}")
    ffn = json.loads((export / "config.json").read_text()).get("ffn_type", "moe")
    if (ffn != spec["ffn_type"]) if dense else ffn in ("dense_shared", "dense"):
        raise RuntimeError(f"dflash export {export} has ffn_type {ffn}, the notebook config says {spec.get('ffn_type')}")
    out = {"ARC_DFLASH_DRAFTER": str(export)}
    if not dense:
        out |= {"ARC_DFLASH_KEEP": str(files["keep"][0]), "ARC_DFLASH_CALIB": str(files["calib"][0])}
    if spec.get("draft_vocab"):
        # the drafter's head over a fixed id list (sglang-arc-2 heads.SubsetHead), pinned by sha256
        path = mounted(spec["draft_vocab_dataset"].split("/")[-1]) / spec["draft_vocab"]
        got = sha256_of(path)
        if got != spec["draft_vocab_sha256"]:
            raise RuntimeError(f"dflash draft vocab {path} is {got}, the config pins {spec['draft_vocab_sha256']}")
        out["ARC_DFLASH_DRAFT_VOCAB"] = str(path)
    state["dflash"] = out
    say("DFLASH", json.dumps(out))
    return out


def install_humming():
    """The Humming wheel the packed INT3 experts run on (config ``int3``), laid out as a PYTHONPATH overlay.

    The runtime archive has no humming-kernels; G4 lays the same 0.1.16 wheel beside it
    (work/qwen38fn_eval/g4_provision_int3_pair.py). Unzipped, not pip-installed, so the .dist-info the INT3
    plugin reads its version from comes along. /kaggle/input is read-only and NVRTC refuses the bundled
    native tools without the exec bit, so the overlay lives under WORK and they are chmod'ed there.
    """
    spec = CFG["int3"]
    wheel = mounted_glob(spec["humming_wheel"])
    got = sha256_of(wheel)
    if got != spec["humming_wheel_sha256"]:
        raise RuntimeError(f"{wheel} is {got}, the config pins {spec['humming_wheel_sha256']}")
    overlay = WORK / "humming-overlay"
    if overlay.exists():
        shutil.rmtree(overlay)
    with zipfile.ZipFile(wheel) as archive:
        archive.extractall(overlay)
    native = sorted((overlay / "humming/_native/x86_64").glob("*"))
    if not native:
        raise RuntimeError(f"the humming wheel laid no native tools under {overlay}/humming/_native/x86_64")
    for path in native:
        path.chmod(0o755)
    dist = sorted(overlay.glob("humming_kernels-*.dist-info"))
    if [d.name for d in dist] != [f"humming_kernels-{spec['humming_version']}.dist-info"]:
        raise RuntimeError(f"humming overlay dist-info {[d.name for d in dist]} != {spec['humming_version']}")
    out = {"overlay": str(overlay), "wheel": str(wheel), "native": len(native)}
    say("HUMMING", json.dumps(out))
    return out


def start_server(runtime, model, env, arch_list):
    env = runtime_env(runtime, env, arch_list)
    env.update({k: str(v) for k, v in CFG.get("server_env", {}).items()})
    if CFG.get("int3"):
        if "humming" not in state:
            raise RuntimeError("the config serves packed INT3 experts but install_humming never ran")
        overlay = state["humming"]["overlay"]
        env["PYTHONPATH"] = os.pathsep.join([overlay, env["PYTHONPATH"]])
        env["ARC_HUMMING_DIR"] = overlay
        if CFG["int3"].get("tuning"):
            table = CODE / CFG["int3"]["tuning"]
            got = sha256_of(table)
            if got != CFG["int3"]["tuning_sha256"]:
                raise RuntimeError(f"{table} is {got}, the config pins {CFG['int3']['tuning_sha256']}")
            env["ARC_INT3_TUNING"] = str(table)
            say("INT3 TUNING", str(table))
    if CFG.get("sglang_tree"):
        if "sglang_tree" not in state:
            raise RuntimeError("the config asks for the HiSparse tree but install_sglang_tree never ran")
        env["ARC_HISPARSE_QSA_MANIFEST"] = state["sglang_tree"]["manifest"]
    if CFG.get("dflash"):
        env.update(dflash_env())
    leaked = [k for k in TEST_ONLY_ENV if k in env]
    if leaked:
        raise RuntimeError(f"test-only switches in the server environment: {leaked}")
    # Record-only: the switches the config hands sglang (none secret). Opt-in features that log nothing when on
    # (h20's ARC_GDN_FOLD_IN_VERIFY only refuses a start without its prerequisites) show up here.
    state["server_env"] = {k: env[k] for k in CFG.get("server_env", {})}
    say("SERVER ENV", json.dumps(state["server_env"]))
    caches = jit_cache_dirs(env)
    for d in caches:
        d.mkdir(parents=True, exist_ok=True)
    for key in ("SGLANG_CACHE_DIR", "SGLANG_JIT_CACHE_DIR", "TRITON_CACHE_DIR", "CUDA_CACHE_PATH"):
        if env.get(key):
            pathlib.Path(env[key]).chmod(0o700)  # root's; no game uid has any business there
    cached_before = cache_files(caches)
    interpreter = runtime_interpreter()
    # Packed INT3: the launcher audits the checkpoint headers, installs the arc_int3 loader in this process and
    # (through its sitecustomize) in every spawned worker, then runs sglang.launch_server with the same argv.
    entry = ([str(CODE / "work/qwen38fn_eval/int3_deploy_runtime.py"), "launch"] if CFG.get("int3")
             else ["-m", "sglang.launch_server"])
    argv = [interpreter, *entry, "--model-path", str(model),
            "--served-model-name", CFG["served_model"], "--host", "127.0.0.1",
            "--port", str(CFG["port"]), "--context-length", str(CFG["context_length"]),
            "--max-running-requests", str(CFG["max_requests"]),
            "--mem-fraction-static", str(CFG["mem_fraction"]), *CFG.get("server_args", [])]
    if CFG.get("max_total_tokens"):
        argv += ["--max-total-tokens", str(CFG["max_total_tokens"])]
    log = (WORK / "server.log").open("w")
    proc = subprocess.Popen(argv, stdout=log, stderr=subprocess.STDOUT, start_new_session=True, env=env)
    base = f"http://127.0.0.1:{CFG['port']}"
    begin = time.monotonic()
    memory = MemorySampler()
    memory.start()
    try:
        while time.monotonic() - begin < CFG["startup_timeout"]:
            if proc.poll() is not None:
                raise RuntimeError("server exited: " + (WORK / "server.log").read_text()[-1500:])
            try:
                with urllib.request.urlopen(base + "/health", timeout=3) as response:
                    if response.status == 200:
                        break
            except Exception:
                time.sleep(3)
        else:
            raise RuntimeError("startup deadline")
    finally:
        state["startup_memory"] = memory.stop()
        compiled = sorted(cache_files(caches) - cached_before)
        state["startup_compiled"] = {
            "new_files": len(compiled), "dirs": [str(d) for d in caches],
            "sgl_jit": sorted({pathlib.Path(f).parent.name for f in compiled if "sgl_kernel_jit" in f})[:60],
            "sample": compiled[:30]}
        save()
        say("STARTUP", json.dumps({"seconds": round(time.monotonic() - begin, 1),
                                   "memory": state["startup_memory"],
                                   "compiled_files": state["startup_compiled"]["new_files"]}))
    # codex 0.154 removed wire_api=chat, so /v1/responses is the only wire left --
    # and sglang turns that endpoint off on an init failure with one WARNING line
    # rather than refusing to start. Offline, its optional downloads are a live
    # candidate for exactly that.
    text = (WORK / "server.log").read_text(errors="replace")
    bad = [line for line in text.splitlines() if "Responses API (/v1/responses) disabled" in line]
    if bad:
        raise RuntimeError("sglang disabled the Responses API: " + bad[0][-300:])
    # A served NVFP4 pool is a claim about the startup log, not about the flags
    # that were passed: sglang will fall back and start quite happily. That
    # check used to be imported from install_qsa_nvfp4_kaggle; with the patch
    # stack baked into the archive there is no installer to import, so the
    # markers live in the config beside every other startup-log claim.
    if CFG.get("kv_cache_dtype") == "nvfp4" and not CFG.get("require_log"):
        raise RuntimeError("kv_cache_dtype is nvfp4 but the config asks for no log "
                           "markers, so nothing would notice the pool falling back")
    # Each of these has a silent fallback behind it: QSA can drop back to its
    # fa2 default, and an `auto` MoE runner can land on trtllm-gen, which has no
    # SM120 cubins. Neither says so; both change what was measured.
    for marker in CFG.get("require_log", []):
        if marker not in text:
            # A marker can be absent for two very different reasons: the
            # fallback really happened, or the string moved. 0.5.20 switched to
            # logging `server_args=<resolved_dict()>`, which turned
            # `moe_runner_backend='x'` into `'moe_runner_backend': 'x'` -- and a
            # gate that fails with nothing but the string it wanted costs a
            # whole run to tell those two apart. So bring the evidence.
            stem = max(marker.replace("'", " ").replace(":", " ").split(), key=len)
            near = [line[-300:] for line in text.splitlines() if stem in line][:4]
            raise RuntimeError(f"server log never said {marker!r}; lines mentioning "
                               f"{stem!r}: {near or 'none'}")
    for marker in CFG.get("refuse_log", []):
        if marker in text:
            raise RuntimeError(f"server log said {marker!r}")
    if CFG.get("sglang_tree"):
        state["hisparse_sizing"] = hisparse_startup_checks(text)
        after = cgroup_now()
        state["memory_after_start"] = after
        if after["headroom"] < CFG["min_headroom_after_start_gib"]:
            raise RuntimeError(f"cgroup headroom {after['headroom']} GiB after start < "
                               f"{CFG['min_headroom_after_start_gib']}: something holds memory the plan "
                               f"does not count: {json.dumps(after)}")
    with urllib.request.urlopen(base + "/get_server_info", timeout=30) as response:
        info = json.loads(response.read())
    state["server_info"] = {k: info.get(k) for k in
                            ("max_total_num_tokens", "context_length", "max_running_requests",
                             "kv_cache_dtype", "page_size", "attention_backend")}
    save()
    say("ALLOC", json.dumps(state["server_info"]))
    return proc, base, round(time.monotonic() - begin, 1), env
```

```python
"""The only cell that runs anything. Everything above it only defines names.

**Save & Run and the scored rerun take the same code path.** The launcher is
always in upstream mode: in the rerun the upstream is the organiser's
`gateway:8001`; in Save & Run it is a stand-in -- the official `create_app()`,
competition mode on, over the public corpus oversampled to the rerun's size --
started here. The old arm's Save & Run took a corpus branch the rerun never
takes (journal 2026-09-18: "要驗的東西 Save & Run 驗不到"); this one rehearses the
path that is scored.
"""

rerun = os.environ.get("KAGGLE_IS_COMPETITION_RERUN", "").lower() in ("1", "true")
state["rerun"] = rerun
save()
stage("arm_files", unpack_arm)
if not rerun:
    # Kaggle validates the *selected version's* Save & Run output at submit time
    # and refuses a version that never wrote submission.parquet. In the rerun the
    # gateway writes the real one and a placeholder would overwrite the score.
    import pandas as pd
    placeholder = WORK / "submission.parquet"
    pd.DataFrame([{"row_id": "1_0", "game_id": "1", "end_of_game": True, "score": 1}],
                 columns=["end_of_game", "game_id", "row_id", "score"]).to_parquet(placeholder, index=False)
    state["submission"] = {"placeholder": str(placeholder), "bytes": placeholder.stat().st_size}
    save()
    say("SUBMISSION", json.dumps(state["submission"]))


def close_parent(corpus_src, restore=True):
    """Hide the corpus by closing the directory its read-only mount sits in.

    Asked as the user who would be reading -- root passes every mode check.
    Measured on this image 2026-09-17: every chmod under /kaggle/input is EROFS,
    Landlock is ABI 0, user namespaces are EPERM; closing /kaggle is what works.
    """
    parent = pathlib.Path("/kaggle")
    before = parent.stat().st_mode & 0o7777
    report = {"path": str(parent), "mode_before": oct(before)}
    try:
        parent.chmod(0o700)
    except OSError as error:
        return {**report, "closes": False, "outcome": f"{type(error).__name__}: {error.strerror}"}
    readable = subprocess.run(["runuser", "-u", "nobody", "--", "test", "-r", str(corpus_src)],
                              capture_output=True).returncode == 0
    report.update(outcome="chmod returned", readable_by_nobody=readable, closes=not readable)
    if restore:
        parent.chmod(before)
        report["mode_restored"] = oct(parent.stat().st_mode & 0o7777)
    return report


def oversample(corpus, target):
    """Distinct ids for the same public games, so Save & Run is rerun-sized.

    The copy's version directory gets a **prefix** tag (`c2<hash>`), never a
    suffix: the official gateway decides "this game already has an environment"
    with `key.startswith(game_id)` (arc_agi/scorecard.py has_environment), so a
    suffixed copy `ar25-0c556536x2` that starts first makes `ar25-0c556536`
    unstartable for the rest of the run -- measured on PC 2026-09-25.
    """
    seeds = sorted(v for g in sorted(corpus.iterdir()) if g.is_dir()
                   for v in sorted(g.iterdir()) if v.is_dir())
    made = []
    while len(seeds) + len(made) < target:
        seed = seeds[len(made) % len(seeds)]
        copy = seed.with_name(f"c{len(made) // len(seeds) + 2}{seed.name}")
        shutil.copytree(seed, copy)
        meta = copy / "metadata.json"
        spec = json.loads(meta.read_text())
        spec["game_id"] = f"{copy.parent.name}-{copy.name}"
        meta.write_text(json.dumps(spec, indent=1))
        made.append(spec["game_id"])
    ids = sorted(f"{g.name}-{v.name}" for g in corpus.iterdir() if g.is_dir() for v in g.iterdir() if v.is_dir())
    clash = [(a, b) for a in ids for b in ids if a != b and b.startswith(a)]
    if clash:
        raise RuntimeError(f"oversampled ids where one is a prefix of another: {clash[:3]}")
    return {"seeds": len(seeds), "added": len(made), "total": len(ids), "sample": made[:3]}


def pack_games(games, out):
    """One ``<game>.tar.gz`` per workspace (its record/, notes, scripts, codex rollouts).

    Kaggle serves the output one file per request, about 0.3 s each, and the game
    record writes several small files per step: the 30-min record probe
    (2026-09-26) left tens of thousands and took an hour to download. gzip level 1:
    the PNG/MP4/NPZ are compressed already, the rollouts are text.
    """
    out.mkdir(parents=True, exist_ok=True)

    def skip(info):
        return None if not (info.isfile() or info.isdir() or info.issym()) else info

    def pack(ws):
        with tarfile.open(out / f"{ws.name}.tar.gz", "w:gz", compresslevel=1) as tar:
            tar.add(ws, arcname=ws.name, filter=skip)

    dirs = sorted(d for d in games.iterdir() if d.is_dir()) if games.is_dir() else []
    with ThreadPoolExecutor(8) as pool:
        list(pool.map(pack, dirs))
    return len(dirs)


def start_standin(harness_python, corpus, port):
    """The official gateway, unchanged, as Save & Run's stand-in for gateway:8001."""
    log = (WORK / "standin.log").open("w")
    proc = subprocess.Popen([harness_python, "-m", "arcforge.arc_code.official",
                             "--corpus", str(corpus), "--port", str(port), "--host", "127.0.0.1",
                             "--scorecard-out", str(WORK / "standin-scorecard.json")],
                            # The kernel's env carries MPLBACKEND=module://matplotlib_inline…,
                            # which matplotlib (imported by arc_agi) rejects outside Jupyter:
                            # v1 of this notebook died here, 6 minutes in.
                            env=dict(os.environ, PYTHONPATH=str(CODE / "src"), MPLBACKEND="Agg"),
                            stdout=log, stderr=subprocess.STDOUT, start_new_session=True)
    begin = time.monotonic()
    while time.monotonic() - begin < 300:
        if proc.poll() is not None:
            raise RuntimeError("stand-in gateway exited: " + (WORK / "standin.log").read_text()[-1500:])
        try:
            with urllib.request.urlopen(f"http://127.0.0.1:{port}/api/healthcheck", timeout=3):
                return proc
        except Exception:
            time.sleep(1)
    raise RuntimeError("stand-in gateway never answered /api/healthcheck")


# Where the agent's world lives: outside /kaggle, which the fence closes.
AGENT.mkdir(parents=True, exist_ok=True)
AGENT.chmod(0o711)
state["agent_root"] = str(AGENT)
tools = stage("tools", stage_tools)
model = stage("model", find_model)
prefetching = start_prefetch(model)
runtime = stage("runtime", unpack_runtime)
state["runtime_provenance"] = stage("provenance", lambda: assert_runtime_provenance(runtime))
state["runtime_overlay"] = stage(
    "overlay", lambda: apply_runtime_overlay(runtime, CFG["runtime_overlay"], CFG["codex_dataset"]))
if CFG.get("sglang_tree"):
    state["sglang_tree"] = stage("sglang_tree", lambda: install_sglang_tree(runtime))
if CFG.get("int3"):
    state["humming"] = stage("humming", install_humming)
wheels = mounted("arc_agi_3_wheels")
corpus_src = mounted("environment_files")
state["mounts"] = {"wheels": str(wheels), "corpus": str(corpus_src)}
save()
venvs = stage("venv", lambda: make_venvs(wheels))
# The corpus copy is the answer key: root only. The launcher reads it for the
# baselines (Save & Run's cross-check) and the stand-in serves it.
corpus = WORK / "environment_files"
if not rerun:
    if not corpus.exists():
        shutil.copytree(corpus_src, corpus)
    corpus.chmod(0o700)
    if CFG.get("oversample_to"):
        state["oversample"] = oversample(corpus, CFG["oversample_to"])
        say("OVERSAMPLE", json.dumps(state["oversample"]))
state["prefetch"] = stage("prefetch", lambda: (PENDING.clear(), join_prefetch(prefetching))[1])
cuda = stage("cuda", lambda: cuda_report(runtime_interpreter(), runtime_env(runtime, os.environ.copy())))
state["rewarm"] = stage("rewarm", lambda: rewarm(model))
server, base, startup, env = stage(
    "server", lambda: start_server(runtime, model, os.environ.copy(), cuda["arch_list"]))[:4]
state["startup_s"] = startup
save()


def snapshot_metrics(name):
    """sglang's /metrics as text into WORK: two of them bracket the arm, so the generated tokens per second
    over the whole arm (prefill stalls included) come from the server's own counters, as on the G4."""
    with urllib.request.urlopen(base + "/metrics", timeout=30) as response:
        (WORK / name).write_bytes(response.read())
    return {"file": name, "at_s": round(time.monotonic() - T0, 1)}


state["metrics_start"] = snapshot_metrics("metrics-start.txt")


class ServerDeathReport(threading.Thread):
    """Report-only: no restart (v11 showed a restart does not finish on Kaggle). When sglang exits during the
    arm, print why -- the tail of its log and the container's memory -- because a Save & Run's live log is
    the only place that is readable while it runs, and the rest of the run is lost either way."""

    def __init__(self, proc):
        super().__init__(daemon=True, name="server-death-report")
        self.proc, self.halt, self.event = proc, threading.Event(), None

    def run(self):
        while not self.halt.wait(10):
            code = self.proc.poll()
            if code is None:
                continue
            self.event = {"exit": code, "at_s": round(time.monotonic() - T0), "memory": cgroup_now()}
            say("SERVER DIED", json.dumps(self.event))
            print((WORK / "server.log").read_text(errors="replace")[-4000:], flush=True)
            return

    def stop(self):
        self.halt.set()
        self.join(timeout=30)


def cgroup_memory() -> dict:
    """This container's memory as the kernel accounts it, in GiB."""
    rel = Path("/proc/self/cgroup").read_text().strip().splitlines()[-1].split(":", 2)[2]
    base = Path("/sys/fs/cgroup") / rel.lstrip("/")
    stat = dict(line.split() for line in (base / "memory.stat").read_text().splitlines())
    out = {k: round(int(stat[k]) / 2**30, 1) for k in ("anon", "shmem", "file", "file_dirty", "file_writeback")}
    out["current"] = round(int((base / "memory.current").read_text()) / 2**30, 1)
    out["max"] = (base / "memory.max").read_text().strip()
    return out


def copy_checkpoint(source: Path, dest: Path, workers: int, chunk: int = 64 << 20,
                    flush_every: int = 1 << 30, report_every: int = 10 << 30) -> dict:
    """The checkpoint on the local disk, for a server restart to load from.

    A restart during the arm cannot load from /kaggle/input: the v11 probe v4
    (2026-09-27) read the checkpoint twice at 9.5 and 27 GB/s right before the
    restart, and the loader still took 14 minutes over its 6th shard of 20 --
    with the arm's own memory in the cgroup the 106 GB no longer fits beside the
    server's host memory, so the mmap loader refaults it off NFS. A refault off
    the local disk is not an NFS round trip.

    v5 copied with shutil on 16 threads and the session died without output a few
    minutes in (the stand-in's 1 s start took 194 s beside it). So now: few
    threads, and every GiB written is synced and dropped from the cache, source
    and copy alike, so dirty pages stay bounded; every 10 GiB prints the cgroup's
    memory and the disk, so the next death says whether it was memory or disk.
    """
    begin = time.monotonic()
    files = sorted(p for p in source.rglob("*") if p.is_file())
    total = sum(p.stat().st_size for p in files)
    free = shutil.disk_usage(dest.parent).free
    if free < total + (20 << 30):
        raise RuntimeError(f"{dest.parent} has {free / 2**30:.0f} GiB free for a {total / 2**30:.0f} GiB copy")
    dest.mkdir(mode=0o700)
    lock, done = threading.Lock(), {"bytes": 0, "reported": 0}

    def one(path):
        target = dest / path.relative_to(source)
        target.parent.mkdir(mode=0o700, parents=True, exist_ok=True)
        src = os.open(path, os.O_RDONLY)
        dst = os.open(target, os.O_WRONLY | os.O_CREAT | os.O_EXCL, 0o600)
        try:
            offset = flushed = 0
            while True:
                block = os.read(src, chunk)
                if not block:
                    break
                view = memoryview(block)
                while view:
                    view = view[os.write(dst, view):]
                offset += len(block)
                if offset - flushed >= flush_every:
                    os.fdatasync(dst)
                    os.posix_fadvise(dst, flushed, offset - flushed, os.POSIX_FADV_DONTNEED)
                    os.posix_fadvise(src, flushed, offset - flushed, os.POSIX_FADV_DONTNEED)
                    flushed = offset
                with lock:
                    done["bytes"] += len(block)
                    if done["bytes"] - done["reported"] >= report_every:
                        done["reported"] = done["bytes"]
                        say("MODEL COPY", f"{done['bytes'] / 2**30:.0f}/{total / 2**30:.0f} GiB in "
                            f"{time.monotonic() - begin:.0f} s; memory", json.dumps(cgroup_memory()),
                            f"disk free {shutil.disk_usage(dest).free / 2**30:.0f} GiB")
            os.fsync(dst)
            os.posix_fadvise(dst, 0, 0, os.POSIX_FADV_DONTNEED)
            os.posix_fadvise(src, 0, 0, os.POSIX_FADV_DONTNEED)
        finally:
            os.close(src)
            os.close(dst)
        if offset != path.stat().st_size:
            raise RuntimeError(f"short copy: {target} {offset} of {path.stat().st_size}")

    with ThreadPoolExecutor(max_workers=workers) as pool:
        for future in as_completed([pool.submit(one, p) for p in files]):
            future.result()
    seconds = time.monotonic() - begin
    return {"dest": str(dest), "files": len(files), "bytes": total, "seconds": round(seconds, 1),
            "GB_per_second": round(total / seconds / 1e9, 3), "memory_after": cgroup_memory()}


local_copy = {}


def run_copy():
    try:
        say("MODEL COPY", "memory before", json.dumps(cgroup_memory()))
        local_copy["report"] = copy_checkpoint(Path(model), Path(CFG["restart_model_copy"]),
                                               CFG.get("model_copy_workers", 2))
        say("MODEL COPY", json.dumps(local_copy["report"]))
    except BaseException as error:  # noqa: BLE001 -- printed now; a restart raises it again
        local_copy["error"] = f"{type(error).__name__}: {error}"[:600]
        say("MODEL COPY FAILED", local_copy["error"])


copying = None
if CFG.get("restart_model_copy"):
    copying = threading.Thread(target=run_copy, daemon=True, name="model-copy")
    copying.start()
    say("MODEL COPY", "started to", CFG["restart_model_copy"])

arm_config = json.loads((UNIT / "configs" / (CFG["rerun_arm_config"] if rerun
                                             else CFG["saverun_arm_config"])).read_text())
served_context = (state.get("server_info") or {}).get("context_length")
if served_context != CFG["context_length"]:
    raise RuntimeError(f"served context {served_context} != notebook config {CFG['context_length']}")
if not arm_config["auto_compact_token_limit"] < arm_config["model_context_window"] <= served_context:
    raise RuntimeError(f"need compact < codex window <= served context: "
                       f"{arm_config['auto_compact_token_limit']} / {arm_config['model_context_window']} / {served_context}")
if CFG.get("codex_tokenizer_sha256"):
    # codex 0007: tool output is cut at tool_output_token_limit *real* tokens of the served
    # model. The game uid cannot reach /kaggle once the parent is closed, so the agent tree
    # (AGENT is 0711) gets a pinned copy, handed to codex through driver_files.
    tokenizer = AGENT / "tokenizer.json"
    shutil.copyfile(pathlib.Path(model) / "tokenizer.json", tokenizer)
    tokenizer.chmod(0o644)
    digest = hashlib.sha256(tokenizer.read_bytes()).hexdigest()
    if digest != CFG["codex_tokenizer_sha256"]:
        raise RuntimeError(f"tokenizer.json sha256 {digest} != notebook config {CFG['codex_tokenizer_sha256']}")
    if "tool_output_tokenizer" in (arm_config.get("driver_files") or {}):
        raise RuntimeError("arm config already sets driver_files.tool_output_tokenizer")
    arm_config.setdefault("driver_files", {})["tool_output_tokenizer"] = str(tokenizer)
    say("TOKENIZER", str(tokenizer), digest[:16])

standin = None
if rerun:
    upstream = CFG["gateway_url"]
    # The same resolution every scored arc-code submission used: the organiser's
    # key if the environment carries one, else the SDK's documented default.
    # The official server does not authenticate; the key only has to be consistent.
    upstream_key = os.environ.get("ARC_API_KEY") or "test-key-123"
    state["upstream_key_source"] = "ARC_API_KEY" if os.environ.get("ARC_API_KEY") else "default test-key-123"
else:
    standin = stage("standin", lambda: start_standin(venvs["harness"], corpus, CFG["standin_port"]))
    upstream = f"http://127.0.0.1:{CFG['standin_port']}"
    upstream_key = "test-key-123"
key_file = WORK / "upstream_key"
key_file.write_text(upstream_key + "\n")
key_file.chmod(0o600)
state["upstream"] = upstream
save()

# Kaggle stops the session nine hours after it started -- before this cell began
# -- and a killed session commits nothing. Five minutes at each end.
# kaggle_session_seconds overrides the nine hours: a timeout test only (does the scored rerun really stop at 9 h?).
KAGGLE_SESSION = CFG.get("kaggle_session_seconds", 32400)
FRONT = 300
BACK = 300
budget = int(KAGGLE_SESSION - FRONT - BACK - (time.monotonic() - T0))
wall = min(arm_config["wall_seconds"], budget)
state["budget"] = {"spent_s": round(time.monotonic() - T0, 1), "playable_s": budget,
                   "configured_wall_s": arm_config["wall_seconds"], "wall_s": wall}
say("BUDGET", json.dumps(state["budget"]))
if wall < 1200:
    raise RuntimeError("less than 20 minutes of playable wall left: " + json.dumps(state["budget"]))
arm_config_path = WORK / "arm-config.json"
arm_config_path.write_text(json.dumps(arm_config, indent=1))
state["arm_config"] = arm_config
save()

# The empty sitecustomize the fenced sessions import instead of the image's.
pyshadow = AGENT / "pyshadow"
pyshadow.mkdir(exist_ok=True)
(pyshadow / "sitecustomize.py").write_text("# Deliberately empty: shadows the host's sitecustomize.\n")
pyshadow.chmod(0o755)
(pyshadow / "sitecustomize.py").chmod(0o644)


def start_model_proxy(spec, image_key, png_size):
    """work/arc_code/sampling_proxy.py between codex and sglang (config model_proxy; arm act_image).

    act_image / act_library end a command's output with signed [[arc-image:...]] markers; the proxy swaps them
    for the PNGs of the game record (act_images.py) before the request reaches sglang, the same bytes every
    time, so the QSA session prefix still resumes. It writes config model_proxy.sampling into every request:
    v19 wrote what the server uses anyway when codex sends none (ResponsesRequest 0.7 / 1.0, top_k unset -> -1
    under --sampling-defaults openai), v30 writes 1.0 / 0.95 / top_k 20 with no act_image or reviewer at all.
    Every other path (/close_session, /metrics, /health, /get_server_info) is relayed byte for byte.
    """
    sets = [a for k, v in sorted(spec["sampling"].items()) for a in ("--set", f"{k}={json.dumps(v)}")]
    argv = [venvs["harness"], "-u", str(UNIT / "sampling_proxy.py"), "--listen", str(spec["port"]),
            "--upstream", base, *sets, "--manifest", str(WORK / "sampling_proxy.json"),
            "--usage-log", str(WORK / "proxy-usage.jsonl")]
    if image_key is not None:
        argv += ["--image-ref-key", str(image_key), "--image-run-dir", str(RUN), "--image-png-size", str(png_size)]
    if arm_config.get("reviewer"):
        # the read-only reviewer (review_scheduler.py): the report lands in review/Rnn.md and the proxy appends a
        # one-line notice to the player's next tool output; run_arm refuses to start without this proxy's marker
        argv += ["--review-run-dir", str(RUN)]
    log = (WORK / "sampling_proxy.log").open("a")
    proc = subprocess.Popen(argv, env=dict(os.environ, PYTHONPATH=str(CODE / "src"), MPLBACKEND="Agg"),
                            stdout=log, stderr=subprocess.STDOUT, start_new_session=True)
    url = f"http://127.0.0.1:{spec['port']}"
    begin = time.monotonic()
    while time.monotonic() - begin < 120:
        if proc.poll() is not None:
            raise RuntimeError("model proxy exited: " + (WORK / "sampling_proxy.log").read_text()[-1500:])
        try:
            # relayed to sglang's /health: a 200 here is the proxy and the server both answering
            with urllib.request.urlopen(url + "/health", timeout=5) as response:
                if response.status == 200:
                    return proc, url
        except Exception:
            time.sleep(1)
    raise RuntimeError("model proxy never relayed /health")


class ProxyKeeper(threading.Thread):
    """Every model request of every game goes through the proxy: dead, it is all 110 games stalled until the
    wall ends. It holds nothing a restart loses but its counters and its image cache, so start it again with
    the same arguments -- loudly, at most max_restarts times."""

    def __init__(self, proc, restart, max_restarts):
        super().__init__(daemon=True, name="proxy-keeper")
        self.proc, self.restart, self.max_restarts = proc, restart, max_restarts
        self.halt, self.events = threading.Event(), []

    def run(self):
        while not self.halt.wait(5):
            code = self.proc.poll()
            if code is None:
                continue
            event = {"exit": code, "at_s": round(time.monotonic() - T0)}
            self.events.append(event)
            say("MODEL PROXY DIED", json.dumps(event))
            print((WORK / "sampling_proxy.log").read_text(errors="replace")[-3000:], flush=True)
            if len(self.events) > self.max_restarts:
                say("MODEL PROXY", f"no restarts left ({self.max_restarts}); the games stall from here")
                return
            try:
                self.proc = self.restart()[0]
                event["restarted"] = True
                say("MODEL PROXY", "restarted, pid", self.proc.pid)
            except BaseException as error:  # noqa: BLE001 -- printed; the next poll sees the dead process again
                event["restart_error"] = f"{type(error).__name__}: {error}"[:300]
                say("MODEL PROXY RESTART FAILED", event["restart_error"])

    def stop(self):
        self.halt.set()
        self.join(timeout=30)


model_base = base
image_key = proxy = proxy_keeper = None
if arm_config.get("reviewer") and not CFG.get("model_proxy"):
    raise RuntimeError("the arm config has a reviewer: its notices reach the players through model_proxy")
png_size = None
if CFG.get("model_proxy"):
    if not (arm_config.get("act_image") or arm_config.get("reviewer")) and \
            CFG["model_proxy"]["sampling"] == {"temperature": 0.7, "top_p": 1.0, "top_k": -1}:
        raise RuntimeError("model_proxy without act_image or a reviewer is only for sampling other than the server's own")
    if arm_config.get("act_image"):
        # middleware signs the image refs, the proxy checks them; root only (/kaggle is closed during the arm)
        image_key = WORK / "image_ref.key"
        image_key.write_text(os.urandom(32).hex() + "\n")
        image_key.chmod(0o600)
        png_size = arm_config["record"]["png_size"]
    proxy, model_base = stage("model_proxy", lambda: start_model_proxy(CFG["model_proxy"], image_key, png_size))
    state["model_proxy"] = {"url": model_base, "pid": proxy.pid, "png_size": png_size,
                            "sampling": CFG["model_proxy"]["sampling"]}
    say("MODEL PROXY", json.dumps(state["model_proxy"]))
    save()
elif arm_config.get("act_image"):
    raise RuntimeError("the arm config turns act_image on: its markers need model_proxy to become images")

argv = [venvs["harness"], "-u", str(UNIT / "run_arm.py"),
        "--config", str(arm_config_path), "--run-dir", str(RUN),
        "--upstream", upstream, "--upstream-key-file", str(key_file),
        "--prep", tools["prep"], "--python", venvs["harness"], "--agent-python", venvs["agent"],
        "--driver-bin", tools["prefix"], "--driver", "codex",
        "--model-base-url", model_base + "/v1", "--model", CFG["served_model"],
        "--wall-seconds", str(wall),
        "--fence-deny", ":".join(str(p) for p in (corpus, corpus_src.parent)),
        "--pyshadow", str(pyshadow), "--tag", CFG.get("kernel_slug", ""),
        # the heartbeat counts QSA session decisions, thinking-budget closes and host KV from it
        "--server-log", str(WORK / "server.log")]
if image_key is not None:
    argv += ["--image-ref-key", str(image_key)]
if not rerun:
    # Save & Run: the launcher cross-checks the stand-in's card against its own
    # ledger RHAE, which needs the baselines.
    argv += ["--corpus", str(corpus)]
# No runtime PYTHONPATH: run_arm is stdlib + arcforge, and the sglang tree on
# sys.path could shadow it.
arm_env = dict(os.environ, PYTHONPATH="", MPLBACKEND="Agg")

# The last thing before the arm: from here no unprivileged uid can traverse into
# /kaggle at all. fence_posix.sh checks that, per game, as the game's own user.
closing = close_parent(corpus_src, restore=False)
state["closing"] = closing
say("CLOSED", json.dumps(closing))
if not closing["closes"]:
    raise RuntimeError("closing /kaggle did not hide the corpus: " + json.dumps(closing))
save()
#: launcher lines worth showing live in the Kaggle log. Everything goes to arm.log;
#: these also go to the notebook's stdout, which is the only thing readable while
#: the session runs (the old arm's log was silent for the whole nine hours).
LIVE = ("HEARTBEAT", "WATCHDOG", "schedule:", "scorecard", "posix fence", "arcgame:", "prompts:", "WIN", "FENCE", "OOM", "CLOSE",
        "official scorecard", "拋出", "放棄", "unit result", "Traceback", "Error", "===", "mean ",
        "MODEL ERROR", "model server", "REVIEW")


class ServerWatchdog(threading.Thread):
    """sglang dead -> start it again with the same arguments and the same log checks.

    The scheduler turns any exception (a CUDA OOM in the vision encoder is the one we
    know of: the ViT runs a whole video at once, outside the static pool) into
    SIGQUIT for the whole server, and nothing restarted it: with the model gone the
    agents failed until their games were retired. codex waits out a refused
    connection on its own (unbounded_connection_retries), so a server that comes
    back within a few minutes costs a pause, not the run. The dead server's log is
    kept as server.<n>.log and its tail printed -- in a Save & Run it is the cause.
    """

    def __init__(self, proc, restart, max_restarts, kill_after=None):
        super().__init__(daemon=True, name="server-watchdog")
        self.proc, self.restart, self.max_restarts = proc, restart, max_restarts
        self.kill_after = kill_after          # Save & Run test: kill the server once, this many s in
        self.smi = shutil.which("nvidia-smi")
        if self.smi is None:
            raise RuntimeError("the server watchdog needs nvidia-smi to see the GPU freed before a restart")
        self.events = []
        self.halt = threading.Event()

    def run(self):
        began = time.monotonic()
        up_since = began
        while not self.halt.wait(5):
            if self.kill_after is not None and time.monotonic() - began >= self.kill_after:
                say("SERVER TEST KILL", f"killing sglang {self.kill_after} s into the arm, as configured")
                os.killpg(self.proc.pid, 9)
                self.kill_after = None
            code = self.proc.poll()
            if code is None:
                continue
            number = len(self.events) + 1
            tail = (WORK / "server.log").read_text(errors="replace")[-2500:]
            (WORK / "server.log").rename(WORK / f"server.{number}.log")
            event = {"n": number, "exit": code, "uptime_s": round(time.monotonic() - up_since),
                     "log": f"server.{number}.log"}
            say("SERVER DIED", json.dumps(event))
            print(tail, flush=True)
            if number > self.max_restarts:
                event["restarted"] = False
                self.events.append(event)
                say("SERVER NOT RESTARTED", f"{self.max_restarts} restarts used")
                return
            try:
                os.killpg(self.proc.pid, 9)
            except ProcessLookupError:
                pass
            say("SERVER RESTART", f"attempt {number} of {self.max_restarts}")
            try:
                event["gpu_used_mib"] = self.wait_gpu_free()
                proc, _, seconds, _ = self.restart()
            except BaseException as error:  # noqa: BLE001 -- recorded and printed; the arm sees a dead server
                event["restart_error"] = f"{type(error).__name__}: {error}"[:600]
                self.events.append(event)
                say("SERVER RESTART FAILED", event["restart_error"])
                return
            self.proc, up_since = proc, time.monotonic()
            event["restart_s"] = seconds
            self.events.append(event)
            say("SERVER RESTARTED", json.dumps(event))

    def wait_gpu_free(self, limit_mib=2048, timeout=180):
        """The dead scheduler's memory goes back when its processes are gone; loading
        the model before that is an OOM at load time."""
        deadline = time.monotonic() + timeout
        while True:
            out = subprocess.run([self.smi, "--query-gpu=memory.used", "--format=csv,noheader,nounits"],
                                 capture_output=True, text=True, check=True).stdout
            used = int(out.split()[0])
            if used < limit_mib:
                return used
            if time.monotonic() > deadline:
                raise RuntimeError(f"GPU still has {used} MiB in use {timeout} s after the server died")
            time.sleep(3)

    def stop(self):
        self.halt.set()
        self.join(timeout=30)


def run_live(argv, timeout, env, log):
    """run(), but the launcher's stdout is read as it is written.

    The timeout is a timer of its own, not a check between lines: a launcher that
    hangs *silently* never yields another line, and a check in the read loop would
    then never run -- the session would sit there until Kaggle's own limit and
    commit nothing. On expiry the launcher's whole process group is killed.
    """
    begin = time.monotonic()
    expired = threading.Event()
    with (WORK / log).open("w") as sink:
        proc = subprocess.Popen(argv, env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                                text=True, bufsize=1, start_new_session=True)

        def expire():
            expired.set()
            print(f"[{time.strftime('%H:%M:%S')}] ARM TIMEOUT after {timeout} s: killing the launcher", flush=True)
            os.killpg(proc.pid, 9)

        timer = threading.Timer(timeout, expire)
        timer.daemon = True
        timer.start()
        try:
            for line in proc.stdout:
                sink.write(line)
                sink.flush()
                if any(key in line for key in LIVE):
                    print(line.rstrip()[:600], flush=True)
            code = proc.wait()
        finally:
            timer.cancel()
    if expired.is_set():
        raise RuntimeError(f"arm exceeded its {timeout} s timeout")
    if code:
        raise RuntimeError(f"arm exited {code}: " + (WORK / log).read_text(errors="replace")[-3000:])
    return {"seconds": round(time.monotonic() - begin, 1)}


def restart_server():
    """What the first start did, in the same order: the checkpoint read twice, then
    sglang -- but from the local copy (copy_checkpoint), which it waits for.

    v2 of the v11 probe restarted straight off /kaggle/input and sat at shard 0/20
    for 20 minutes; v4 read it twice first and still crawled (see copy_checkpoint).
    The first read here is the local disk's speed, measured, not assumed.
    """
    if copying is None:
        raise RuntimeError("no restart_model_copy configured: a restart off /kaggle/input "
                           "does not finish (v11 probe v2/v4)")
    say("SERVER RESTART", "waiting for the model copy")
    copying.join()
    if "error" in local_copy:
        raise RuntimeError("model copy failed: " + local_copy["error"])
    local = local_copy["report"]["dest"]
    for label in ("disk", "warm"):
        began = time.monotonic()
        manifest = prefetch_checkpoint(Path(local), CFG.get("prefetch_workers", 16))
        say("SERVER RESTART", f"checkpoint read ({label}) in {time.monotonic() - began:.0f} s,",
            f"{manifest['GB_per_second']:.2f} GB/s")
    say("SERVER RESTART", "starting sglang on", local)
    return start_server(runtime, local, os.environ.copy(), cuda["arch_list"])


# The watchdog is a probe, not part of a submission (2026-09-27): its restart
# does not finish on Kaggle -- off /kaggle/input the loader crawls, a local copy
# trips the session's disk limit -- and the gateway closes an idle scorecard after
# 15 minutes anyway. Without "server_watchdog" the arm runs as v10 did.
watchdog = None
if CFG.get("server_watchdog"):
    watchdog = ServerWatchdog(
        server, restart_server,
        max_restarts=CFG["server_max_restarts"],
        # The test kill is for Save & Run only; a scored rerun never kills its own server.
        kill_after=None if rerun else CFG.get("test_kill_server_after_s"))
    watchdog.start()
elif CFG.get("test_kill_server_after_s") or CFG.get("restart_model_copy") or "server_max_restarts" in CFG:
    raise RuntimeError("test kill / model copy / server_max_restarts configured without server_watchdog")
death_report = None
if watchdog is None:
    death_report = ServerDeathReport(server)
    death_report.start()
if proxy is not None:
    proxy_keeper = ProxyKeeper(proxy, lambda: start_model_proxy(CFG["model_proxy"], image_key, png_size),
                               CFG["model_proxy"]["max_restarts"])
    proxy_keeper.start()
try:
    stage("arm", lambda: run_live(argv, wall + BACK, arm_env, "arm.log"))
finally:
    if watchdog is not None:
        watchdog.stop()
        server = watchdog.proc
        state["server_restarts"] = watchdog.events
    if death_report is not None:
        death_report.stop()
        state["server_death"] = death_report.event
    if proxy_keeper is not None:
        proxy_keeper.stop()
        proxy = proxy_keeper.proc
        state["model_proxy"]["deaths"] = proxy_keeper.events
    if server.poll() is None:
        try:
            state["metrics_end"] = snapshot_metrics("metrics-end.txt")
        except BaseException as error:  # noqa: BLE001 -- teardown: recorded and printed; the evidence below matters more
            state["metrics_end"] = {"error": f"{type(error).__name__}: {error}"[:300]}
            say("METRICS END FAILED", state["metrics_end"]["error"])
    state["memory_at_end"] = cgroup_now()
    if copying is not None:
        copying.join()
    state["model_copy"] = local_copy
    for proc in (server, standin, proxy):
        if proc is not None and proc.poll() is None:
            os.killpg(proc.pid, 15)
    # /kaggle is reopened for the teardown: the evidence goes back into WORK.
    pathlib.Path("/kaggle").chmod(int(closing["mode_before"], 8))
    arm_state = RUN / "ARM_STATE.json"
    if arm_state.exists():
        state["arm"] = json.loads(arm_state.read_text())

    def not_copyable(directory, names):
        # Unix sockets and fifos cannot be copied ([Errno 6]); losing the whole
        # run directory to that is losing the evidence the run exists to produce.
        # The game workspaces are packed separately (pack_games). A reviewer's snapshot (reviews/<game>/Rnn/ws)
        # hardlinks the game record: copied out it is a second copy of every frame -- its record/ stays behind,
        # and in the rerun (nobody sees the output) the whole snapshot does; Rnn/REVIEW.md and result.json go.
        here = pathlib.Path(directory)
        snapshot = here.name == "ws" and here.parent.parent.parent == RUN / "reviews"
        return [n for n in names
                if stat.S_ISSOCK(os.lstat(os.path.join(directory, n)).st_mode)
                or stat.S_ISFIFO(os.lstat(os.path.join(directory, n)).st_mode)
                or here == RUN / "games"
                or snapshot and (rerun or n == "record")]

    try:
        if RUN.exists():
            shutil.copytree(RUN, WORK / "arm-run", symlinks=True, dirs_exist_ok=True,
                            ignore=not_copyable, ignore_dangling_symlinks=True)
            # Nobody sees the rerun's output; packing 110 workspaces there only eats
            # into the margin between the arm and the session's end.
            packed = 0 if rerun else pack_games(RUN / "games", WORK / "arm-run" / "games")
            state["arm_run_materialised"] = {"packed_games": packed, "bytes": sum(
                f.lstat().st_size for f in (WORK / "arm-run").rglob("*") if f.is_file() and not f.is_symlink())}
    except BaseException as error:
        state["arm_run_materialised"] = {"error": f"{type(error).__name__}: {error}"[:600]}
    # The per-game tokens and the competition key are not evidence, and v3's
    # output carried both. Dropped from the committed copy.
    for secret in ("gateway/tokens", "gateway/upstream_key"):
        (WORK / "arm-run" / secret).unlink(missing_ok=True)
    say("MATERIALISE", json.dumps(state.get("arm_run_materialised")))
    text = (WORK / "arm.log").read_text(errors="replace") if (WORK / "arm.log").exists() else ""
    if text.strip():
        say("ARM_TAIL", text[-6000:])
    dropped = {}
    for name in ("runtime", "environment_files"):
        path = WORK / name
        if path.is_dir() and not path.is_symlink():
            try:
                dropped[name] = sum(f.lstat().st_size for f in path.rglob("*") if f.is_file() and not f.is_symlink())
                shutil.rmtree(path)
            except BaseException as error:
                dropped[name] = f"{type(error).__name__}: {error}"[:300]
    for name in ("upstream_key", "image_ref.key"):
        (WORK / name).unlink(missing_ok=True)
    state["dropped_from_working"] = dropped
    try:
        if AGENT.is_dir():
            shutil.rmtree(AGENT)
        state["dropped_agent_root"] = str(AGENT)
    except BaseException as error:
        state["dropped_agent_root"] = f"{type(error).__name__}: {error}"[:300]
    state["complete"] = True
    save()
say("DONE", disk())
final = (WORK / "arm-run" / "FINAL.txt")
if final.exists():
    say("FINAL\n" + final.read_text())
arm_state = state.get("arm") or {}
say("RESULT", json.dumps({k: arm_state.get(k) for k in ("status", "error", "schedule", "watchdog")})[:4000])
```
